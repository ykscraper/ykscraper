```python

import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import csv

url = "https://www.codewithharry.com/courses"

response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

courses_main_title = soup.find("title")
print(courses_main_title.get_text(strip=True))

print(soup.title)

courses = soup.select('h2.text-xl.font-semibold.text-primary.mb-2')
course_prices = soup.select("span.text-lg.font-bold.text-primary")

course_links = [
    course.find_parent("a")
    for course in courses
]

course_images = soup.select("img.absolute.top-0.left-0.w-full.h-full.object-cover")

with open("courses.csv", "w", newline="", encoding="utf-8-sig") as file:
    writer = csv.writer(file)

    writer.writerow(["Course Name", "Price", "Course URL", "Image URL"])

    for course, course_price, course_link, course_image in zip(
        courses, course_prices, course_links, course_images
    ):
        name = course.get_text(strip=True)
        price = course_price.get_text(strip=True)
        link = urljoin(url, course_link.get("href"))
        image = urljoin(url, course_image.get("src"))

        print(name, price, link, image)

        writer.writerow([name, price, link, image])

print("Data saved to courses.csv")

import pandas as pd
from openpyxl import load_workbook
from openpyxl.styles import Font

df = pd.read_csv("courses.csv")

df.to_excel("courses.xlsx", index=False)

wb = load_workbook("courses.xlsx")
ws = wb.active

for row in range(2, ws.max_row + 1):
    for col in [3, 4]:
        cell = ws.cell(row=row, column=col)

        if cell.value and str(cell.value).startswith("http"):
            cell.hyperlink = cell.value
            cell.font = Font(color="0563C1", underline="single")

wb.save("courses.xlsx")

print("Excel file created successfully!")
from google.colab import files
files.download("courses.xlsx")

```
